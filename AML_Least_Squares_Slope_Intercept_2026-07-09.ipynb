### Calculate the slope and intercept using the least squares method

For a simple linear regression line $y = mx + b$, the slope ($m$) and y-intercept ($b$) can be calculated using the following formulas:

```python
#question 1 to determine the best fit line for a small dataset using the least square method in simple linear regression
import numpy as np
import matplotlib.pyplot as plt
X=np.array([1,2,3,4])
Y=np.array([2,3,4,5])
x_mean=np.mean(X)
y_mean=np.mean(Y)
numerator=np.sum(X-x_mean)*(Y-y_mean)
denominator=np.sum((X-x_mean)**2)
m=numerator/denominator
b=y_mean-m*x_mean
print("Slope:",m)
print("Intercept:",b)


# EQUation of the line is
print("Equation of the line is y=",m,"x+",b)

```

### Q2. Determine the linear relationship between variables using the least squares method

We will use the data `x=[1,2,3,4,5]` and `y=[6,7,8,9,10]` from the previous cell (`GevNnO6NrxX4`) to perform simple linear regression using the least squares method.

```python
import numpy as np

# Ensure x and y are numpy arrays for calculations
x = np.array(x)
y = np.array(y)

n = len(x)

# Calculate sums required for the formulas
sum_x = np.sum(x)
sum_y = np.sum(y)
sum_xy = np.sum(x * y)
sum_x_squared = np.sum(x**2)

# Calculate slope (m) using the formula: m = (n*sum(xy) - sum(x)*sum(y)) / (n*sum(x^2) - (sum(x))^2)
m = (n * sum_xy - sum_x * sum_y) / (n * sum_x_squared - sum_x**2)

# Calculate intercept (b) using the formula: b = (sum(y) - m*sum(x)) / n
b = (sum_y - m * sum_x) / n

print(f"Calculated Slope (m): {m:.4f}")
print(f"Calculated Intercept (b): {b:.4f}")
print(f"The equation of the best fit line is: y = {m:.2f}x + {b:.2f}")
```

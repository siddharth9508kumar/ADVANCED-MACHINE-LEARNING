# one way anova


```python
import numpy as np
from scipy.stats import f_oneway
# Marks obtained by students
programmer_A = np.array([70, 72, 68])
programmer_B = np.array([80, 82, 78])
programmer_C = np.array([90, 88, 92])
# Perform One-Way ANOVA
F_value, p_value = f_oneway(
    programmer_A,
    programmer_B,
    programmer_C
)
print("F-value:", F_value)
print("p-value:", p_value)
# Significance level
alpha = 0.05
if p_value < alpha:
    print("Reject H0")
    print("There is a significant difference between the group means.")
else:
    print("Fail to reject H0")
    print("There is no significant difference between the group means.")
```

# two way anova


```python
import pandas as pd
data = pd.DataFrame({
    'Study_Time': ['Low', 'Low', 'High', 'High'],
    'Method': ['A', 'B', 'A', 'B'],
    'Marks': [60, 70, 80, 90]
})
print("Dataset:")

print(data)
#calculate Means
grand_mean = data['Marks'].mean()
method_mean = data.groupby('Method')['Marks'].mean()
study_mean = data.groupby('Study_Time')['Marks'].mean()
print("\nGrand Mean:", grand_mean)
print("\nMethod Means:")
print(method_mean)
print("\nStudy Time Means:")
print(study_mean)
# Sum of Squares for Method
n_method = 2
SS_method = n_method * sum(
    (method_mean[m] - grand_mean) ** 2
    for m in method_mean.index
)
# Sum of Squares for Study Time
n_study = 2
SS_study = n_study * sum(
    (study_mean[s] - grand_mean) ** 2
    for s in study_mean.index
)
# Total Sum of Squares
SS_total = sum(
    (x - grand_mean) ** 2
    for x in data['Marks']
)
# Interaction Sum of Squares
SS_interaction = 0
for study in data['Study_Time'].unique():
    for method in data['Method'].unique():
        cell_value = data[
            (data['Study_Time'] == study) &
            (data['Method'] == method)
        ]['Marks'].iloc[0]
        expected_value = (
            study_mean[study]
            + method_mean[method]
            - grand_mean
        )
        SS_interaction += (
            cell_value - expected_value
        ) ** 2
# Error Sum of Squares
SS_error = SS_total - SS_method - SS_study - SS_interaction
# Degrees of Freedom

df_method = 1
df_study = 1
df_interaction = 1
df_error = 0
df_total = 3
# Display results
print("\nANOVA Calculations")
print("SS Method =", SS_method)
print("SS Study Time =", SS_study)
print("SS Interaction =", SS_interaction)
print("SS Error =", SS_error)
print("SS Total =", SS_total)
print("\nDegrees of Freedom")
print("Method =", df_method)
print("Study Time =", df_study)
print("Interaction =", df_interaction)
print("Error =", df_error)
print("Total =", df_total)
print("\nImportant:")
print("F-value and p-value cannot be calculated")
print("because Error degrees of freedom = 0.")
```
